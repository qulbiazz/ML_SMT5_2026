```python
import numpy as np
import pandas as pd

df = pd.read_csv('../../data/spam.csv', encoding='latin-1') # spesifiksi encoding diperlukan karena data tidak menggunakan UTF-8

df.head()
```

```python

# Mengubah label numerik menjadi label teks
df['Labels'] = df['Labels'].map({
    0: 'ham',
    1: 'spam'
})

# Cek jumlah data per kelas
print(df['Labels'].value_counts())
print('\n')

# Cek kelengkapan data
df.info()
print('\n')

# Cek statistik deskriptif
print(df.describe())
```

```python
# Data untuk label
new_labels = {
    'spam': 1,
    'ham': 0
}

# Encode label
df['Labels'] = df['Labels'].map(new_labels)

# Cek data
df.head()
```

```python
X = df['SMS'].values
y = df['Labels'].values
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer

# Pastikan X berupa teks SMS dan y berupa label
X = df['SMS'].astype(str)
y = df['Labels']

# Split data training dan testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=50
)

# Inisiasi CountVectorizer
bow = CountVectorizer()

# Fit dan transform data training
X_train = bow.fit_transform(X_train)

# Transform data testing menggunakan vocabulary training
X_test = bow.transform(X_test)

print("Ukuran X_train:", X_train.shape)
print("Ukuran X_test:", X_test.shape)
print("Ukuran y_train:", y_train.shape)
print("Ukuran y_test:", y_test.shape)
```

```python

# Menghitung jumlah fitur atau kosakata
print(len(bow.get_feature_names_out()))

# Menampilkan dimensi data training
print(f'Dimensi data: {X_train.shape}')
```

```python

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Inisiasi MultinomialNB
mnb = MultinomialNB()

# Fit model dengan data training
mnb.fit(X_train, y_train)

# Prediksi data training
y_pred_train = mnb.predict(X_train)

# Evaluasi akurasi data training
acc_train = accuracy_score(y_train, y_pred_train)

# Prediksi data testing
y_pred_test = mnb.predict(X_test)

# Evaluasi akurasi data testing
acc_test = accuracy_score(y_test, y_pred_test)

# Print hasil evaluasi
print(f'Hasil akurasi data train: {acc_train}')
print(f'Hasil akurasi data test: {acc_test}')
```
