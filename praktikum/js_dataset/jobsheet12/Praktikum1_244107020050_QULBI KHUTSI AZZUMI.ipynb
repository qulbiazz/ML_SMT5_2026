```python
import numpy as np

# Dataset XOR
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([[0], [1], [1], [0]])

# Fungsi aktivasi
def sigmoid(x):
    x = np.clip(x, -500, 500)
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(a):
    return a * (1 - a)

def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x > 0).astype(float)


# Fungsi training ANN
def train_ann(hidden_size, activation, epochs=10000, lr=0.1):

    np.random.seed(42)

    # Inisialisasi bobot
    W1 = np.random.randn(2, hidden_size)
    b1 = np.zeros((1, hidden_size))

    W2 = np.random.randn(hidden_size, 1)
    b2 = np.zeros((1, 1))

    for epoch in range(epochs):

        # Forward pass
        z1 = np.dot(X, W1) + b1

        if activation == "relu":
            a1 = relu(z1)
        else:
            a1 = sigmoid(z1)

        z2 = np.dot(a1, W2) + b2
        a2 = sigmoid(z2)

        # Menghitung error
        error = y - a2

        # Backpropagation output layer
        d_a2 = error * sigmoid_derivative(a2)

        d_W2 = np.dot(a1.T, d_a2)
        d_b2 = np.sum(d_a2, axis=0, keepdims=True)

        # Backpropagation hidden layer
        d_a1 = np.dot(d_a2, W2.T)

        if activation == "relu":
            d_z1 = d_a1 * relu_derivative(z1)
        else:
            d_z1 = d_a1 * sigmoid_derivative(a1)

        d_W1 = np.dot(X.T, d_z1)
        d_b1 = np.sum(d_z1, axis=0, keepdims=True)

        # Update bobot
        W1 += lr * d_W1
        b1 += lr * d_b1

        W2 += lr * d_W2
        b2 += lr * d_b2

        # Menampilkan loss
        if epoch % 1000 == 0:
            loss = np.mean(error ** 2)
            print(f"Epoch {epoch}, Loss: {loss:.6f}")

    # Prediksi akhir setelah bobot diperbarui
    z1 = np.dot(X, W1) + b1

    if activation == "relu":
        a1 = relu(z1)
    else:
        a1 = sigmoid(z1)

    a2 = sigmoid(np.dot(a1, W2) + b2)

    final_loss = np.mean((y - a2) ** 2)

    return final_loss, a2


# Percobaan 1: Konfigurasi awal
print("\n=== PERCOBAAN 1 ===")
loss1, pred1 = train_ann(2, "sigmoid")

# Percobaan 2: Hidden layer 3 neuron
print("\n=== PERCOBAAN 2 ===")
loss2, pred2 = train_ann(3, "sigmoid")

# Percobaan 3: Hidden layer 3 neuron + ReLU
print("\n=== PERCOBAAN 3 ===")
loss3, pred3 = train_ann(3, "relu")


# Perbandingan hasil
print("\n=== PERBANDINGAN LOSS ===")
print(f"2 Neuron + Sigmoid : {loss1:.6f}")
print(f"3 Neuron + Sigmoid : {loss2:.6f}")
print(f"3 Neuron + ReLU    : {loss3:.6f}")

print("\nPrediksi 3 Neuron + ReLU:")
print(np.round(pred3, 4))
```
