## Klasifikasi Topik Artikel Klarifikasi Hoaks Komdigi dengan SVM

Notebook ini mengganti regresi gaji yang tidak sesuai dengan dataset menjadi klasifikasi teks. Fitur berasal dari `title`, `excerpt`, dan `body_text`; targetnya adalah `topics`. `topics` merupakan kategori artikel, bukan label benar atau salahnya klaim.


```python
# Import package
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, f1_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.svm import LinearSVC
```

```python
# Cari dataset dari direktori kerja notebook hingga root proyek.
data_path = next(
    (
        parent / "data" / "komdigi_hoaks.csv"
        for parent in (Path.cwd(), *Path.cwd().parents)
        if (parent / "data" / "komdigi_hoaks.csv").is_file()
    ),
    None,
)
if data_path is None:
    raise FileNotFoundError("File data/komdigi_hoaks.csv tidak ditemukan dari direktori kerja saat ini.")

data = pd.read_csv(data_path)
required_columns = ["url", "title", "excerpt", "body_text", "topics"]
missing_columns = [column for column in required_columns if column not in data.columns]
if missing_columns:
    raise ValueError(f"Kolom wajib tidak ditemukan pada dataset: {missing_columns}")

# Hapus artikel duplikat dan siapkan teks yang akan diklasifikasikan.
data = data.drop_duplicates(subset="url", keep="first").dropna(subset=["topics"]).copy()
text_columns = ["title", "excerpt", "body_text"]
data[text_columns] = data[text_columns].fillna("").astype(str)
data["article_text"] = (
    data[text_columns]
    .agg(" ".join, axis=1)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
data = data.loc[data["article_text"].ne("")].reset_index(drop=True)

print(f"Lokasi dataset: {data_path.resolve()}")
print("Ukuran dataset setelah menghapus duplikat:", data.shape)
display(data[["topics", "title", "excerpt"]].head())
print("Distribusi kategori topics:")
display(data["topics"].value_counts().rename_axis("topics").to_frame("jumlah"))
```

```python
# Visualisasi distribusi kelas untuk melihat ketidakseimbangan label.
plt.figure(figsize=(10, 5))
class_order = data["topics"].value_counts().index
sns.countplot(data=data, y="topics", order=class_order, color="steelblue")
plt.title("Distribusi Topik Artikel Komdigi")
plt.xlabel("Jumlah artikel")
plt.ylabel("Topik")
plt.tight_layout()
plt.show()
```

```python
# Artikel menjadi fitur teks dan topics menjadi target klasifikasi.
X = data["article_text"]
y = data["topics"]

# Bagi data 75:25 dengan menjaga proporsi setiap kategori.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y,
)
print("Jumlah data latih:", len(X_train))
print("Jumlah data uji:", len(X_test))
```

```python
# TF-IDF mengubah artikel menjadi fitur numerik untuk SVM linear.
# class_weight="balanced" membantu kategori minoritas saat pelatihan.
model = Pipeline(
    [
        (
            "tfidf",
            TfidfVectorizer(
                max_features=10000,
                min_df=2,
                ngram_range=(1, 2),
                sublinear_tf=True,
            ),
        ),
        (
            "classifier",
            LinearSVC(class_weight="balanced", random_state=42),
        ),
    ]
)
model.fit(X_train, y_train)
print("Model SVM linear selesai dilatih.")
```

```python
# Ukur performa pada data uji yang tidak digunakan saat pelatihan.
y_pred = model.predict(X_test)
labels = sorted(y.unique())
print("Akurasi data uji:", round(accuracy_score(y_test, y_pred), 4))
print("Macro F1-score:", round(f1_score(y_test, y_pred, average="macro", zero_division=0), 4))
print("\nLaporan klasifikasi:")
print(
    classification_report(
        y_test,
        y_pred,
        labels=labels,
        target_names=labels,
        zero_division=0,
    )
)
```

```python
# Matriks kebingungan memperlihatkan kategori yang sering tertukar.
cm = confusion_matrix(y_test, y_pred, labels=labels)
plt.figure(figsize=(9, 7))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=labels,
    yticklabels=labels,
)
plt.title("Confusion Matrix - SVM Topik Artikel Komdigi")
plt.xlabel("Prediksi")
plt.ylabel("Label sebenarnya")
plt.xticks(rotation=35, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Tinjau prediksi model pada beberapa artikel di data uji.
hasil_prediksi = data.loc[X_test.index, ["title", "topics"]].rename(
    columns={"topics": "topics_asli"}
).copy()
hasil_prediksi["topics_prediksi"] = y_pred
display(hasil_prediksi.head(10))
```
